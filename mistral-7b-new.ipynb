# **Custom HF Embeddings (transformers-only)**

```python
# =========================
# Custom HF Embeddings (GPU-optimized, FP16)
# =========================
from typing import List
import torch
from transformers import AutoTokenizer, AutoModel
from langchain.embeddings.base import Embeddings  # <-- fixed import

class HFTransformersEmbeddings(Embeddings):
    """
    GPU-optimized embeddings using HuggingFace transformers.
    Mean-pools last_hidden_state. Supports FP16 for faster GPU inference.
    """
    def __init__(self, model_name: str, device: str = "cpu", normalize: bool = True, use_fp16: bool = True):
        self.device = device
        self.normalize = normalize
        self.tokenizer = AutoTokenizer.from_pretrained(model_name, use_fast=True)
        
        self.model = AutoModel.from_pretrained(model_name)
        
        # Move model to device
        self.model.to(self.device)
        
        # Optional FP16 for GPU
        if use_fp16 and self.device == "cuda":
            self.model.half()
        
        self.model.eval()
        print(f"Embedding model loaded on {self.device}, FP16={use_fp16 and self.device=='cuda'}")

    @torch.no_grad()
    def _embed(self, texts: List[str]) -> List[List[float]]:
        # Tokenize and move to device
        enc = self.tokenizer(
            texts,
            padding=True,
            truncation=True,
            max_length=512,
            return_tensors="pt"
        ).to(self.device)

        # Forward pass
        outputs = self.model(**enc)
        last_hidden = outputs.last_hidden_state  # [B, T, H]
        attn_mask = enc["attention_mask"].unsqueeze(-1)  # [B, T, 1]

        # Mean pooling
        masked = last_hidden * attn_mask
        sums = masked.sum(dim=1)
        counts = attn_mask.sum(dim=1).clamp(min=1)
        embs = sums / counts

        # Normalize embeddings
        if self.normalize:
            embs = torch.nn.functional.normalize(embs, p=2, dim=1)

        return embs.cpu().tolist()

    def embed_documents(self, texts: List[str]) -> List[List[float]]:
        return self._embed(texts)

    def embed_query(self, text: str) -> List[float]:
        return self._embed([text])[0]
```

# **Load Mistral‑7B‑Instruct v0**

```python
import torch
from transformers import (
    AutoTokenizer,
    AutoModelForCausalLM,
    pipeline,
    BitsAndBytesConfig
)
from langchain_community.llms import HuggingFacePipeline

# =========================
# Detect GPU
# =========================
DEVICE = "cuda" if torch.cuda.is_available() else "cpu"
print(f"Using device: {DEVICE}")

MODEL_NAME = "mistralai/Mistral-7B-Instruct-v0.3"

# =========================
# Load tokenizer
# =========================
tokenizer = AutoTokenizer.from_pretrained(
    MODEL_NAME,
    use_fast=True
)

if tokenizer.pad_token is None:
    tokenizer.pad_token = tokenizer.eos_token

tokenizer.padding_side = "left"

# =========================
# Load model (4-bit GPU if possible)
# =========================
if DEVICE == "cuda":
    try:
        print("Attempting 4-bit GPU quantization...")

        bnb_config = BitsAndBytesConfig(
            load_in_4bit=True,
            bnb_4bit_use_double_quant=True,
            bnb_4bit_quant_type="nf4",
            bnb_4bit_compute_dtype=torch.float16,
        )

        model = AutoModelForCausalLM.from_pretrained(
            MODEL_NAME,
            quantization_config=bnb_config,
            device_map="auto",
            torch_dtype=torch.float16,
        )

        print("SUCCESS: Loaded Mistral in 4-bit on GPU")

    except Exception as e:
        print("4-bit failed:", e)
        print("Falling back to FP16 GPU")

        model = AutoModelForCausalLM.from_pretrained(
            MODEL_NAME,
            torch_dtype=torch.float16,
            device_map="auto",
        )
else:
    print("No GPU detected. Loading on CPU")
    model = AutoModelForCausalLM.from_pretrained(
        MODEL_NAME,
        torch_dtype=torch.float32,
        device_map={"": "cpu"},
    )

# =========================
# Optimize memory
# =========================
model.config.use_cache = False
model.eval()

# =========================
# Create generation pipeline
# =========================
gen_pipe = pipeline(
    "text-generation",
    model=model,
    tokenizer=tokenizer,
    do_sample=False,
    repetition_penalty=1.05,
    return_full_text=False,
    pad_token_id=tokenizer.pad_token_id,
    eos_token_id=tokenizer.eos_token_id,
    max_new_tokens=256,
)

# =========================
# LangChain wrapper
# =========================
llm = HuggingFacePipeline(pipeline=gen_pipe)

print("SUCCESS: Mistral-7B-Instruct ready for FAISS RAG")
```

# **Multi-Query Retriever**

```python
from langchain_core.prompts import PromptTemplate
from langchain.retrievers.multi_query import MultiQueryRetriever

# =========================
# Multi-Query Prompt (optimized for Mistral)
# =========================
MULTI_QUERY_PROMPT = PromptTemplate(
    input_variables=["question"],
    template=(
        "You are an expert retrieval assistant.\n"
        "Generate exactly 3 alternative versions of the user's question.\n"
        "These will be used to retrieve relevant information from a vector database.\n\n"
        "Rules:\n"
        "- Keep meaning the same\n"
        "- Use different wording\n"
        "- Be concise\n"
        "- Output ONLY the 3 questions separated by newline\n\n"
        "Original question:\n"
        "{question}"
    ),
)

# =========================
# Base retriever (FAISS)
# =========================
base_retriever = vector_db.as_retriever(
    search_kwargs={
        "k": 3  # number of chunks per query
    }
)

# =========================
# Multi-query retriever using Mistral LLM
# =========================
multi_retriever = MultiQueryRetriever.from_llm(
    retriever=base_retriever,
    llm=llm,  # HuggingFacePipeline wrapper
    prompt=MULTI_QUERY_PROMPT,
    include_original=True  # VERY IMPORTANT (improves retrieval quality)
)

print("Multi-query retriever ready (FAISS + Mistral)")
```

# **RAG Prompt**

```python
from langchain_core.prompts import PromptTemplate

RAG_PROMPT = PromptTemplate(
    input_variables=["chat_history", "context", "question"],
    template=(
        "You are a helpful AI assistant.\n\n"
        "Use ONLY the provided context to answer the question.\n"
        "Use conversation history when it is relevant.\n"
        "Do NOT use outside knowledge.\n"
        "If the answer is not found in the context, respond exactly with:\n"
        "\"I don't know based on the provided document.\"\n\n"
        "---------------------\n"
        "Conversation History:\n"
        "{chat_history}\n\n"
        "---------------------\n"
        "Context:\n"
        "{context}\n\n"
        "---------------------\n"
        "Question:\n"
        "{question}\n\n"
        "Answer:"
    )
)
```

# **Token budgeted context**

```python
from langchain.memory import ConversationBufferWindowMemory
from IPython.display import Markdown, display
import torch
import gc

# -------------------------
# Initialize windowed memory (prevents token overflow)
# -------------------------
memory = ConversationBufferWindowMemory(
    memory_key="chat_history",
    input_key="input",
    output_key="output",
    return_messages=False,
    k=1  # remember last 5 interactions only
)

print("Conversation memory initialized (windowed)")

# -------------------------
# Chat function with memory + FAISS
# -------------------------
def chat_with_documents(question, doc_name=None, budget_tokens=3000):
    """
    Chat over documents using FAISS retrieval + memory.
    """

    # Clean memory & GPU cache
    gc.collect()
    if torch.cuda.is_available():
        torch.cuda.empty_cache()

    # -------------------------
    # Step 1: Retrieve context
    # -------------------------
    context = build_context(
        question=question,
        doc_name=doc_name,
        budget_tokens=budget_tokens
    )

    # -------------------------
    # Step 2: Load memory safely
    # -------------------------
    chat_history = memory.load_memory_variables({}).get("chat_history", "")

    # -------------------------
    # Step 3: Format RAG prompt
    # -------------------------
    prompt = RAG_PROMPT.format(
        chat_history=chat_history,
        context=context,
        question=question
    )

    # -------------------------
    # Step 4: Run LLM (cleaned output)
    # -------------------------
    output = llm(prompt)

    if isinstance(output, list):  # safety for pipeline edge cases
        output = output[0]

    output = output.strip()

    # -------------------------
    # Step 5: Save to memory
    # -------------------------
    memory.save_context(
        {"input": question},
        {"output": output}
    )

    # -------------------------
    # Step 6: Display answer
    # -------------------------
    display(Markdown(output))

    return output

```

# **Final chain**

```python
def fit_context_by_tokens(texts, tokenizer, budget_tokens=3000):
    """
    Fit multiple document texts into a token budget.

    - Uses tokenizer to count tokens per chunk.
    - Stops adding when budget exceeded.
    """
    kept, total = [], 0
    for t in texts:
        # Use tokenizer.encode with truncation=False to count tokens
        n = len(tokenizer.encode(t, truncation=False))
        if total + n > budget_tokens:
            break
        kept.append(t)
        total += n
    return "\n\n".join(kept)


def build_context(question, doc_name=None, budget_tokens=3000, top_k=3):
    """
    Retrieve context from FAISS / multi-query retriever and fit into token budget.

    Args:
        question (str): user question
        doc_name (str, optional): if provided, restrict retrieval to that document
        budget_tokens (int): max tokens for context
        top_k (int): number of top documents to retrieve

    Returns:
        str: combined document context fitting within token budget
    """
    # -------------------------
    # Step 1: Retrieve relevant docs
    # -------------------------
    if doc_name:
        # Filter retrieval to a specific document
        retriever = vector_db.as_retriever(
            search_kwargs={"k": top_k, "filter": {"source": doc_name}}
        )
        docs = retriever.get_relevant_documents(question)
    else:
        # Multi-query retriever across all documents
        docs = multi_retriever.get_relevant_documents(question)

    # -------------------------
    # Step 2: Extract text
    # -------------------------
    texts = [d.page_content for d in docs]

    # -------------------------
    # Step 3: Fit into token budget
    # -------------------------
    context = fit_context_by_tokens(texts, tokenizer, budget_tokens=budget_tokens)
    return context
```
